```python
import datetime

# Get the current datetime object
now = datetime.datetime.now()
date_format = now.strftime('%m-%d-%Y')

time_raw = now.strftime('%H:%M:%S')
time_format = time_raw.replace(':', '%3A')

print(f"Current Date (MM-DD-YYYY): {date_format}")
print(f"Current Time (HH%3AMM%3ASS): {time_format}")
```
